```python
import pandas as pd
```

```python
df.columns = [col.replace('Smooth KDE Bootstrap', '_T4') for col in df.columns]
```

```python
df.columns = [col.replace('Smooth KDE ', '_T4') for col in df.columns]
```

```python
df.columns = [col.replace('KDE Bootstrap', '_T3') for col in df.columns]
df.columns = [col.replace('KDE ', '_T3') for col in df.columns]
```

```python
df
```

```python
phrase = 'Normal TI Coverage'
df = df.loc[:, ~df.columns.str.contains(phrase, case=False)]
```

```python
df
```

```python
df.columns = [col.replace('Bootstrap', '_T2') for col in df.columns]
df.columns = [col.replace('Bootstrap', '_T2') for col in df.columns]
```

```python
df.columns = [col.replace('Non Parametric', '_T1') for col in df.columns]
df.columns = [col.replace('Non Parametric', '_T1') for col in df.columns]
```

```python
df.columns = [col.replace('Coverages', '_O1') for col in df.columns]
df.columns = [col.replace('Widths', '_O2') for col in df.columns]
```

```python
df.columns = [col.replace(' ', '') for col in df.columns]
df.columns = [col.replace('TI', '') for col in df.columns]
```

```python
df.columns = [col.replace('Normal', 'B1_') for col in df.columns]
```

```python
df.columns = [col.replace('Chi-Square', 'B2_') for col in df.columns]
```

```python
df.columns = [col.replace('FDistribution', 'B3') for col in df.columns]
df.columns = [col.replace('LogB1', 'B4') for col in df.columns]
df.columns = [col.replace('TDistribution', 'B5') for col in df.columns]
df.columns = [col.replace('Gamma', 'B6_') for col in df.columns]
df.columns = [col.replace('Beta', 'B7_') for col in df.columns]
df.columns = [col.replace('Exponential', 'B8_') for col in df.columns]
df.columns = [col.replace('Pareto', 'B9_') for col in df.columns]
```

```python
df.columns
```

```python
df.columns = [col.replace('Distribution_', '') for col in df.columns]
```

```python
df
```

```python
def reorder_first_six(col):
    if len(col) < 6:
        return col  # leave short column names as-is
    first3 = col[:3]
    next3 = col[3:6]
    rest = col[6:]
    return next3 + first3 + rest

# Apply to all column names
df.columns = [reorder_first_six(col) for col in df.columns]
```

```python
df.to_csv('final.csv')
```
